```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math as math
```

Dwie metody służące do sortowania na Serii:
- sort_values (na wartości)
- sort_index (na indeksie)

Metody te przyjmuje parametry:
- ascending=True
- inplace=False (dodatkowo tworzy nowy obiekt)

Metody `sort_values()` i `sort_index()` sortują odpowiednio po wartościach i po indeksie serii. Domyślnie `ascending=True`; bez `inplace=True` metody zwracają nową serię, a oryginalny obiekt (`pok`) pozostaje niezmieniony.

```python
pok = pd.read_csv("/content/pokemon.csv", usecols=['Name'], squeeze=True)
```

```python
pok.head()
```

```python
pok.sort_values().head()
```

```python
pok.sort_values(ascending=False).head()
```

```python
# Samo sortowanie nie wpływa na obiekt pok

pok.head()
```

```python
# Aby zapisać sortowanie stosuje parametr

pok.sort_values(inplace=True)
pok.head()
```

```python
# Sortowanie po indeksie 

pok.sort_index().head()
```

```python
# Sortowanie po indeksie malejąco (tworzę nową Serię)

pok = pok.sort_index(ascending=False)
pok.head()
```
