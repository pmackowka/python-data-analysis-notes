```python
import pandas as pd
import numpy as np
```

Metoda `transpose()` (lub `.T`) zamienia wiersze z kolumnami — przydatne, gdy dane wygodniej przegląda się w układzie obróconym o 90 stopni, np. gdy kolumn jest niewiele, a wierszy dużo.

```python
incidents = pd.read_csv('/content/Canadian Railway Crossing Incidents.csv')
incidents.set_index(['Region', 'EventType'],inplace=True)
incidents.sort_index(inplace=True)
incidents.head(10)
```

```python
incidents.count()
```

```python
len(incidents)
```

```python
incidents.transpose()
```

```python
events = incidents.transpose()
events
```
