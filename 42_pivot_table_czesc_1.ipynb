```python
import pandas as pd
import numpy as np
```

Budowa tabeli przestawnej: `df.pivot_table(index=..., columns=..., values=..., aggfunc=...)` tworzy tabelę z wielopoziomowym indeksem/kolumnami, którą można dalej porządkować metodami `swaplevel()` i `sort_index()`.

```python
sales = pd.read_csv('/content/WA_Sales_Products_2012-14.csv')
sales.head()
```

Pełna sygnatura `pivot_table()`: `values=None, index=None, columns=None, aggfunc='mean', fill_value=None, margins=False, dropna=True, margins_name='All'` — domyślnym sposobem agregacji jest średnia (`aggfunc='mean'`), a `margins=True` dodaje wiersz/kolumnę z podsumowaniem.

```python
sales.pivot_table(index='Retailer country', columns=['Year','Quarter'],values='Revenue').head()
```

```python
sales.pivot_table(index='Retailer country', columns=['Year'],values='Revenue').head()
```

```python
sales.pivot_table(index='Retailer country', columns='Year', values='Gross margin', aggfunc='min').head()
```

```python
sales.pivot_table(index='Retailer country', columns='Year', values='Gross margin', aggfunc=['min', 'max']).head()
```

```python
pivTab = sales.pivot_table(index='Retailer country', columns='Year', values='Gross margin', aggfunc=['min', 'max'])
pivTab.columns
```

```python
pivTab = pivTab.swaplevel(axis='columns').sort_index(axis='columns').head()
pivTab.head()
```
