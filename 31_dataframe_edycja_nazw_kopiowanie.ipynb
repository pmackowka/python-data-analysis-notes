Nazwy kolumn można nadpisać zbiorczo (`frame.columns = [...]`) albo zmienić wybrane etykiety — zarówno kolumn, jak i wartości indeksu — metodą `rename(columns={...})` / `rename(index={...})`. Ważna pułapka: `frameCopy = frame` nie tworzy kopii, tylko drugą referencję do tego samego obiektu w pamięci — zmiana w `frameCopy` zmienia też `frame`. Prawdziwą, niezależną kopię daje dopiero `frame.copy()`.

```python
frame = pd.read_csv('/content/mammals.csv')
frame.head()
```

```python
frame.columns
```

```python
frame.columns = ['name', 'bodyKg', 'brainKg']
frame.head()
```

Jeśli mam dużo kolumn stosuje słownik aby podmienić punktowo nazwy

```python
newColumnNames = {'bodyKg':'body_kg','brainKg':'brain_kg'}
frame.rename(columns = newColumnNames, inplace=True)
frame.head()
```

```python
frame = pd.read_csv('/content/mammals.csv',index_col='name')
frame.head()
```

```python
frame.rename(index = {'Cow':'Land cow'}, inplace=True)
frame.head()
```

```python
frameCopy = frame
frameCopy.head()
```

```python
frameCopy.loc['Arctic fox', 'body'] = 5
frameCopy.head()
```

frameCopy - to tylko wskaźnik w pamięci do oryginalnego DataFrame

```python
frame.head()
```

Teraz kopiuje za pomocą copy()

```python
frameCopy = frame.copy()
frameCopy.loc['Grey wolf', 'body'] = 40
frameCopy.head()
```
