##append i concat

Metoda `append()` (usunięta w nowszych wersjach pandas na rzecz `pd.concat()`) oraz `pd.concat([df1, df2, ...])` sklejają DataFrame'y jeden pod drugim. Parametr `ignore_index=True` nadaje wynikowi nowy, ciągły indeks zamiast zachowywać oryginalne indeksy obu ramek.

```python
suppliers = pd.read_csv('/content/suppliers.csv', usecols = ['SupplierID', 'CompanyName', 'Country'])
suppliers.head()
```

```python
customers = pd.read_csv('/content/customers.csv', usecols=['CustomerID', 'CompanyName', 'Country'])
customers.head()
```

```python
len(supplires)
```

```python
len(customers)
```

```python
suppliers.append(customers)
```

```python
df_append = suppliers.append(customers)
len(df_append)
```

```python
suppliers2 = suppliers.copy()
customers2 = customers.copy()
```

```python
suppliers2.rename({ 'SupplierID' : 'CustomerID'}, axis='columns', inplace=True)
```

```python
suppliers2.head()
```

```python
suppliers2.append(customers)
```

```python
df_concat = pd.concat(objs=[suppliers2,customers])
len(df_concat)
```

##Merge()

Metoda `merge()` łączy dwa DataFrame po wspólnej kolumnie lub indeksie, z pełną kontrolą nad sposobem łączenia przez parametr `how`: `"inner"` zostawia tylko wiersze wspólne dla obu tabel, `"left"`/`"right"` zachowują wszystkie wiersze z jednej strony (dopełniając brakujące dane wartością `NaN`), a `"outer"` zachowuje wiersze z obu tabel niezależnie od dopasowania.

```python
cat1 = pd.read_csv('/content/categories_del_1.csv', usecols = ['CategoryID', 'CategoryName'])
cat2 = pd.read_csv('/content/categories_del_2.csv', usecols = ['CategoryID', 'CategoryName'])
```

```python
cat1
```

```python
cat2
```

```python
cat1.merge(cat2)
```

```python
cat1.merge(cat2, on='CategoryID')
```

```python
cat1.merge(cat2, on='CategoryID', suffixes=['_1','_2'])
```

```python
cat1.merge(cat2,on='CategoryID', suffixes=['_1','_2'],how='outer')
```

```python
cat1.merge(cat2,on='CategoryID', suffixes=['_1','_2'],how='outer',indicator=True)
```

```python
cat_merged = cat1.merge(cat2,on='CategoryID', suffixes=['_1','_2'],how='outer',indicator=True)
```

```python
cat_merged['_merge'] == 'left_only'
```

```python
filter = cat_merged['_merge'] == 'left_only'
cat_merged[filter]
```

```python
filter = cat_merged['_merge'] == 'right_only'
cat_merged[filter]
```

```python
filter = cat_merged['_merge'] == 'both'
cat_merged[filter]
```

```python
cat1.merge(cat2,on='CategoryID', suffixes=['_1','_2'],how='left',indicator=True)
```

```python
cat1.merge(cat2,on='CategoryID', suffixes=['_1','_2'],how='right',indicator=True)
```

##Opcje polecenia merge

Najważniejsze opcje `merge()`: `left_on`/`right_on` pozwalają łączyć po kolumnach o różnych nazwach w obu tabelach, `left_index`/`right_index` łączą po indeksie zamiast kolumny, `suffixes` rozstrzyga konflikt nazw kolumn występujących w obu ramkach, a `sort=True` sortuje wynik po kluczu łączenia.

```python
products = pd.read_csv('/content/products.csv', usecols = ['ProductID', 'ProductName', 'CategoryID', 'UnitPrice'])

categories = pd.read_csv('/content/categories1.csv', 
                              usecols = ['CategoryID', 'CategoryName', 'Description'])
```

```python
products.head()
```

```python
categories.head()
```

```python
products.merge(right=categories, on='CategoryID').head()
```

```python
categories.rename({'CategoryID' : 'ID'}, axis='columns', inplace=True)
categories.head()
```

```python
products.merge(right=categories, left_on='CategoryID', right_on='ID').head()
```

```python
products.rename({'ProductName' : 'Name'}, axis='columns', inplace=True)
```

```python
categories.rename({'CategoryName' : 'Name'}, axis='columns', inplace=True)
```

```python
products.head(3)
```

```python
categories.head(3)
```

```python
products.merge(right=categories, left_on='CategoryID', right_on='ID').head()
```

```python
products.merge(right=categories, left_on='CategoryID', right_on='ID', suffixes=['_prod', '_category']).head()
```

```python
products.merge(right=categories, left_on='CategoryID', right_on='ID', suffixes=['_prod', '_category'],
               sort=True).head()
```

```python
categories.head()
```

```python
categories.set_index('ID', inplace=True)
categories.head()
```

```python
products.merge(right=categories, left_on='CategoryID', right_index=True, suffixes=['_prod', '_category'],
               sort=True).head()
```

##Łączenie 3 i więcej Data Frame

Łańcuch wywołań `df1.merge(df2, ...).merge(df3, ...)` pozwala połączyć więcej niż dwie tabele w jednym wyrażeniu — każde kolejne `merge()` działa na wyniku poprzedniego. Warto ustawiać `suffixes` przy każdym łączeniu osobno, żeby uniknąć niejednoznacznych nazw kolumn w wyniku końcowym.

```python
products = pd.read_csv('/content/products.csv', usecols = ['ProductID', 'ProductName', 'CategoryID', 'UnitPrice'])

categories = pd.read_csv('/content/categories1.csv', usecols = ['CategoryID', 'CategoryName', 'Description'])

orders = pd.read_csv('/content/order-details.csv')
```

```python
products.head(2)
```

```python
categories.head(2)
```

```python
orders.head(2)
```

```python
cat_prod = products.merge(categories, on='CategoryID')
cat_prod.head(2)
```

```python
cat_prod.merge(orders, on='ProductID').head()
```

```python
cat_prod.merge(orders, on='ProductID', suffixes=['_Prod', '_Order']).head()
```
