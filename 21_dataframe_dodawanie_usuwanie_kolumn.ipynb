Nową kolumnę tworzy proste przypisanie do nieistniejącej jeszcze etykiety — `frame['TotalFat0'] = 0` albo z operacji na innych kolumnach: `frame['SugarAndFat'] = frame.Sugars + frame.TotalFat` — taka kolumna trafia zawsze na koniec DataFrame. Metoda `insert(loc=2, column=..., value=...)` pozwala wstawić nową kolumnę w wybranym miejscu zamiast na końcu. Kolumnę usuwa instrukcja `del frame['kolumna']`.

```python
frame = pd.read_csv('/content/Mcdonalds.csv', usecols=['Item', 'Category', 'Calories', 'TotalFat', 'Sugars', 'Protein', 'Cholesterol'])
frame.head()
```

Odwołanie do pojedyńczej kolumny

```python
frame.TotalFat.head()
```

```python
frame['TotalFat'].head()
```

Dodanie nowej kolumny

```python
frame['TotalFat0'] = 0
frame.head()
```

```python
SugarAndFat = frame.Sugars + frame.TotalFat
frame['SugarAndFat'] = SugarAndFat
frame.head()
```

```python
SugarAndFat.head()
```

Dodanie kolumny w jednej instrukcji

```python
frame['SugarAndProtein'] = frame.Sugars + frame.Protein
frame.head()
```

Nowe kolumny dochodzą na sam koniec, a co jeśli chcę umieścić kolumnę gdzieś w środku

```python
frame.insert(loc=2, column='SPF', value=frame.Sugars+frame.Protein+frame.TotalFat)
frame.head()
```

Usuwanie kolumn

```python
del frame['SugarAndProtein']
frame.head()
```
